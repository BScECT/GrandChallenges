The function below counts how many days in a list had rain. Run it, and then try it on a list of your own.

```python
def count_rainy_days(rainfall):
    rainy_days = 0
    for amount in rainfall:
        if amount > 0:
            rainy_days = rainy_days + 1
    return rainy_days

rainfall_week = [0.0, 2.4, 11.2, 0.0, 5.1, 0.5, 1.8]  # in mm per day
print(f"It rained on {count_rainy_days(rainfall_week)} days this week.")
```
