# Introduction to numpy - practice notebook

In this notebook we will introduce numpy in more detail.

Recall that we have been working with lists. Below we create two lists, and add them. What do you think will happen?

```python
x = [2,3,5,7]
y = [1,1,2,3]

print(x+y)
```

Adding two lists, will just stitch them together, which can be convenient... for instance if you want to combine a list with temperatures from two consecutive years.

What if we multiply them?|

```python
print(x*y)
```

Oops, that doesn't work.

But we can multiply a list with a scalar:

```python
z = x*3
print(z)
```

The original list `x` is now repeated 3 times in the new list `z`.

**But what if we want to do some 'proper math' with vectors (or: arrays)?**

Numpy is a Python module for arrays of numbers, and for doing mathematical operations on them.
Numpy is used for data analysis, plotting, linear algebra and is a very commonly used module in scientific Python.

Let's import numpy:


```python
import numpy as np
```

Now functions in numpy are accessible, such as `np.array()`.

The part `as np` is not necessary, but commonly done for slightly shorter code than typing `numpy.array()` every time (just like we imported `matplotlib.pyplot` as `plt` before).


A numpy vector is very much like a Python list, but you can do mathematics on the whole array at once.

We can create a numpy array from a Python list:


```python
# either by 'converting' a previously assigned list
x_list = np.array(x)
print('x_list =',x_list)

# or by just putting a new list inside np.array()
x_new = np.array([5, 2, 3.3])
print('x_new  =',x_new)
```

Note: the outer parentheses `()` belong to the function call. The inner `[]` define a Python list, which is used to provide the values for the vector being created.

There are other ways to create numpy arrays:

```python
z = np.zeros(4) 
e = np.ones(3)
a = np.zeros_like(e)
```

# 6.1.1 Array math and plotting

Now we can do math with the arrays:

```python
y = x_new + e    
print('y       =',y)

print('y*x_new =',y*x_new)
```

Addition and multiplication are done element by element. There is also a _dot product_ which you may know from vectors in mathematics:

```python
a = np.array([ 1, 3, 0, 2])
b = np.array([-1, 2, 2, 0])

c = np.dot(a, b)
print(c)
```

In Python, dot product can also be written as `a@b`, as shown below:

```python
print(a@b)
```

You cannot add arrays of different length, that results in an error:


```python
x1 = np.array([1, 2])
x2 = np.array([1, 2, 3])
print(x1 + x2)
```

But you can add a single number to an array (it adds the number to all elements in the array):

```python
print('x1. =',x1)
print('x1+1=',x1 + 1)
```

Numpy also has a similar function to `range()` which is `np.arange()`:

```python
np.arange(0.5, 4.5, 0.5)
```

Note that like `range()`, the end point is not included. 

If you want the end point to be included robustly, you can instead use `np.linspace(start, end, N)`.
This creates an array of `N` equally spaced numbers where both start and end are included. The step size is determined accordingly.

**Use whichever is more convenient.**

<div class="alert alert-block alert-info">

Note the differences! 

With `np.arange(0, 1, 0.1)` you create an array starting at 0 with step size 0.1 (defined by the last number); the end point is not included: in this case the array has 10 elements `[0, 0.1, 0.2, ..., 0.9]`. 

With `np.linspace(0, 1, 10)` you create an array with 10 elements (defined by the last number), but since in this case the end point is included, it yields `[0, 0.1111111, 0.22222222, ..., 0.88888889, 1]`. 


</div>

Let's now create an array `x` with equally spaced points, and evaluate a function on it.

```python
x1 = np.linspace(0, 20, 100)
y1 = np.exp(x)
```

Note we used `np.exp()` allows to use a numpy array as input, such that the exponential function is evaluated for all elements in the array `x1`.

Use the cell below to print `y1` to check.

```python
### use this cell to print y1
```

Now we can use the arrays to plot the cosine function, using `matplotlib`.


```python
import matplotlib.pyplot as plt
plt.plot(x1, y1,'.')
plt.xlabel('x')
plt.ylabel('exp(x)');                                                                                                                             
```

## 6.1.2 Saving your plots

One more important matplotlib function is saving plots in a file, so that you can use them in reports.


```python
plt.plot(x1, y1,'.-')
plt.xlabel('x')
plt.ylabel('exp(x)') 
plt.grid()
plt.savefig('expplot.png')
```

Matplotlib will now save your plot in the file you specified. The file will be placed in the same folder / directory as your notebook. You can also specify the directory that you would want the image to be saved: check out the documentation of the `savefig` function [here](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.savefig.html). 

Matplotlib can handle several formats: .png (an image), .pdf (vector graphics, small files and often looks good), .svg (also vector graphics, can be edited in for example Inkscape)

[Inkscape](https://inkscape.org) is a free and open-source vector graphics editor, worth learning.
