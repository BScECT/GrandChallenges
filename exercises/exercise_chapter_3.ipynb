## 3.2 If...Else

<div class="alert alert-block alert-info">
<b>Exercise 3.2.1 | If...Elif...Else</b><br>
<br>
<div style="text-align: justify">
Booleans are used to direct the flow of your code using <code>if</code> <code>elif</code> and <code>else</code> statements.
</div>
</div>

The keyword `if` is always followed by a boolean statement and then a `:`. If the boolean statement evaluates as `True` the codeblock that comes next is executed. If it is `False`, it moves to the following `elif` or `else` (if they exist).

```python
if True:
    print("This will always be printed")
else:
    print("This will never be printed")
```

Of course you can (and should) do this with more elaborate boolean statements. Play around with the variables below and see what happens.

```python
the_final_answer = 42

if the_final_answer == 42:
    print("The answer to the ultimate question of life, the universe, and everything.")

elif the_final_answer == 41 or the_final_answer == 43:
    print("The answer is close, but not quite right.")

else:
    print("This cannot be it.")
```
